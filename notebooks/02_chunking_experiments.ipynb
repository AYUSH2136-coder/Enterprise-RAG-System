```python
import os
import sys
sys.path.insert(0, os.path.abspath(".."))

from src.ingestion.text_chunker import chunk_text

# Load the cleaned text we saved in notebook 01
with open("../data/processed/cleaned_text.txt", "r", encoding="utf-8") as f:
    cleaned_text = f.read()

print(f"Loaded cleaned text: {len(cleaned_text)} characters")
```

```python
experiments = [
    {"chunk_size": 200, "chunk_overlap": 40},
    {"chunk_size": 300, "chunk_overlap": 60},
    {"chunk_size": 500, "chunk_overlap": 100},
    {"chunk_size": 800, "chunk_overlap": 150},
    {"chunk_size": 1000, "chunk_overlap": 200},
]

results = []
for exp in experiments:
    chunks = chunk_text(
        cleaned_text,
        chunk_size=exp["chunk_size"],
        chunk_overlap=exp["chunk_overlap"]
    )
    results.append({
        **exp,
        "num_chunks": len(chunks),
        "avg_length": sum(len(c) for c in chunks) // len(chunks),
        "min_length": min(len(c) for c in chunks),
        "max_length": max(len(c) for c in chunks),
    })

# Print comparison table
print(f"{'Size':>6} {'Overlap':>8} {'Chunks':>8} {'Avg Len':>8} {'Min':>6} {'Max':>6}")
print("-" * 50)
for r in results:
    print(f"{r['chunk_size']:>6} {r['chunk_overlap']:>8} "
          f"{r['num_chunks']:>8} {r['avg_length']:>8} "
          f"{r['min_length']:>6} {r['max_length']:>6}")
```

```python
import re

chunks = chunk_text(cleaned_text, chunk_size=500, chunk_overlap=100)

# Check how many chunks end mid-sentence (no period/question/exclamation at end)
incomplete = 0
for chunk in chunks:
    stripped = chunk.strip()
    if stripped and stripped[-1] not in ".?!:\"'":
        incomplete += 1

total = len(chunks)
print(f"Total chunks             : {total}")
print(f"Ending mid-sentence      : {incomplete}")
print(f"Ending at sentence boundary: {total - incomplete}")
print(f"Clean ending %           : {((total - incomplete)/total)*100:.1f}%")
```
